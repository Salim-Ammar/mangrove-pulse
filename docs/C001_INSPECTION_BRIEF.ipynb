# Mangrove Pulse
## Satellite evidence → targeted inspection

Arab Youth Space Hackathon: 813 Challenge · 2 October 2026 evidence snapshot

Mangrove Pulse investigates how satellite evidence can guide targeted mangrove field inspection, with an amphibious drone proposed as a future inspection platform.

We make ambiguous satellite patches inspectable through locations, dated evidence and unanswered questions, rather than treating map disagreement as a diagnosis. Better evidence could help a monitoring team decide what to investigate; operational usefulness remains untested.

| Status | What the team has and intends |
|---|---|
| Completed — RESULT | Historical Khor Kalba Sentinel-2/GMW baseline, spatially held-out evaluation, disagreement/coverage audit, cached reproduction and three review cases; separate GMW reference-sensitivity comparison with WorldCover context. |
| Proposed — HYPOTHESIS | Hand off locations and evidence requests for closer inspection. A drone is a candidate observation platform, not an implemented component. |
| Future research — HYPOTHESIS | Test amphibious access and sensing against simpler methods; consider planting only after establishing ecological need and suitability, then conduct separate mechanism and survival trials. |

The drone is a concept, with no physical prototype or operational tests. A separate team-run Colab study explores hypothetical mission-energy break-even conditions; assumed observation equivalence and power/mass inputs are not verified. It does not prove feasibility or superiority over existing UAE restoration approaches. No planting, survival or carbon benefit is demonstrated.

The figures and evidence are embedded in this reading companion. The [current Markdown brief](C001_INSPECTION_BRIEF.md#acceptance-and-decision-contract--proposed-not-executed) supplies the later proposed acceptance rules and decision outcomes; the [executed main notebook](../Mangrove_Pulse_End_to_End_Executed.ipynb) records the computational demonstration.
